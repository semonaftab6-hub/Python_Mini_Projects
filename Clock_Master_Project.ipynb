```python

# Clock Master Project
# Step 3: Digital Clock

# Import datetime to access the current date and time
from datetime import datetime #Imports a Python class for working with dates and times.

# Import time to pause the program
import time #Imports time-related functions.


# Define a function for the digital clock
def digital_clock():

    # Display the module heading
    print("\n--- DIGITAL CLOCK ---")

    # Keep displaying the time until the user stops the cell
    try:
        while True:

            # Get the current date and time from the system
            current_time = datetime.now()

            # Display the date and time in a readable format =The format "%d-%m-%Y %H:%M:%S" means day-month-year followed by hour-minute-second.
            print(current_time.strftime("%d-%m-%Y %H:%M:%S"))

            # Wait for one second before displaying again,Pauses the program for one second.
            time.sleep(1)

    # Handle the user interrupting the running cell
    except KeyboardInterrupt:
        print("Digital clock stopped.")
```

```python
# Calling the digital clock function
digital_clock()
```

```python
# Step 4: Alarm Clock
# Define the alarm clock function
def alarm_clock():

    # Display the module heading
    print("\n--- ALARM CLOCK ---")

    # Ask the user to enter the alarm time
    # Use the 24-hour format HH:MM, for example 16:30
    alarm_time = input("Enter alarm time (HH:MM): ")

    # Check whether the entered time has the correct format
    try:
        datetime.strptime(alarm_time, "%H:%M")

    # Handle an invalid time format
    except ValueError:
        print("Invalid time! Please use HH:MM format.")
        return

    # Confirm that the alarm has been set
    print("Alarm set successfully!")

    # Continue checking the current time
    while True:

        # Get the current time in hours and minutes
        current_time = datetime.now().strftime("%H:%M")

        # Check whether the current time matches the alarm time
        if current_time == alarm_time:

            # Display the alarm message
            print("\nALARM! Wake up!")

            # Stop the loop after the alarm triggers
            break

        # Wait five seconds before checking again
        time.sleep(5)
```

```python

# Call the alarm clock function
alarm_clock()
```

```python
# Function to create a countdown timer
import time

def countdown_timer():
    # Display the timer heading
    print("\n--- COUNTDOWN TIMER ---")

    # Ask the user to enter time in seconds
    seconds = input("Enter countdown time in seconds: ")

    # Check whether the input contains a valid integer
    try:
        seconds = int(seconds)

        # Check that the entered time is positive
        if seconds <= 0:
            print("Please enter a number greater than zero.")
            return

    except ValueError:
        print("Invalid input! Please enter a whole number.")
        return

    # Start the countdown
    while seconds > 0:
        print("Time remaining:", seconds, "seconds")
        time.sleep(1)
        seconds = seconds - 1

    # Display the completion message
    print("Time is up!")
```

```python
countdown_timer()
```
