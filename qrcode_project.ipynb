```python
import qrcode
url=input("Enter URL:")
if url.startswith("http://")or url.startswith("https://"):
    qr=qrcode.QRCode(version=1,#qrcode ka version set karo , 1 to 40 range  mein version de sakte hain
#qr code ke starting size ko define karta hai ,density set karta hai
                     error_correction=qrcode.constants.ERROR_CORRECT_H,#H=HIGH,L=LOW,M=MEDIUM
                     box_size=10,#10x10 ke boxes qr ke andar
                     border=4
                    )
    qr.add_data(url)#url mein is qr(setting) ko add kardo
    qr.make()
    image=qr.make_image(fill_color="Blue",back_color="red")
    image.save("3rdQR.png")
else:
    print("INVALID URL")
```

```python
#create qr code for upi
import qrcode 
upi_id=input("enter upi id")
name=input("enter name of the person doing payment:")
upi_url=f"upi://pay?pa=(upi_id)&pn=(name)" #string formatting ,creating url,pa =jo upi id se pay horaha hai ,pn = jis parson jo bhejhna hai uska name
qr=qrcode.make(upi_url)
qr.save("semon_UPI.png")
print("UPI KA QR BAN GYA")
```

```python
import qrcode
upi_id=input("enter upi id")
name=input("enter name of the person doing payment:")
amount=int(input("enter amount:"))
upi_url=f"upi://pay?pa=(upi_id)&pn=(name)&am=(amount)" #string formatting ,creating url,pa =jo upi id se pay horaha hai ,pn = jis parson jo bhejhna hai uska name
qr=qrcode.make(upi_url)
qr.save("semon_UPI_2.png")
print("UPI KA QR BAN GYA")
#pa=payee,pn=payee name,am=amount
```

```python
import qrcode
upi_id=input("enter the upi id:")
name=input("enter name:")
upi_url=f"upi://pay?pa=(upi_id)&pn=(name)&am=(amount)"
amount=int(input("enter amount "))
qr=qrcode.QRCode(version=1,error_correction=qrcode.constants.ERROR_CORRECT_H,box_size=10,border=4)
qr.add_data(upi_url)
qr.make()
image=qr.make_image(fill_color="blue",back_color="red")
image.save("semon_upi_3.png")
```
