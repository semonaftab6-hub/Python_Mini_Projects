```python
#project2 
#text to image ........(we need image lib,imagedraw,imagefont in pil module)
from PIL import Image,ImageDraw,ImageFont
#create a blank image canvas
txt=input("enter text:")
img=Image.new("RGB",(800,300),"white") #COLOR COMBINATION,HEIGHT-WIDTH OF FRAME ,BACKGROUND COLOR
#load font or set font
font=ImageFont.load_default()#default font
#draw text on image
draw=ImageDraw.Draw(img)
draw.text((50,100),txt,fill="black",font=font)#50 to left and 100 from top ,font = by default font  
#to save the created image
img.save("txt_image2.png")
print("Image Generated Successfully")
```
